# Diabetes Hospital Readmission — Exploratory Data Analysis

This notebook contains the initial exploration of the data used in our NOVA IMS Data Science Bootcamp project, which aims to predict hospital readmission risk for diabetic patients.

We work with two datasets:
- `diabetic_data.csv`: the main dataset, with ~101,766 hospital encounters and 50 features (patient demographics, admission details, diagnoses, medications, and the readmission outcome).
- `IDS_mapping.csv`: a lookup table that translates the numeric codes used in `diabetic_data.csv` (admission type, discharge disposition, admission source) into their text descriptions.

This section covers: loading both datasets, a first look at their structure and content (shape, preview, data types, descriptive statistics), and an automated profiling report for a more in-depth data quality assessment.

```python
import pandas as pd
from pathlib import Path
from ydata_profiling import ProfileReport
```

```python
import sys

print(sys.executable)
```

```python
# Find the project root, starting from the current working directory
current_dir = Path.cwd()
project_root = current_dir.parent if current_dir.name == "notebooks" else current_dir

# Build paths to the CSV files
data_dir = project_root / "data"
diabetes_path = data_dir / "diabetic_data.csv"
mapping_path = data_dir / "IDS_mapping.csv"

# Load the CSV files
df_diabetes = pd.read_csv(diabetes_path)
df_mapping = pd.read_csv(mapping_path)
```
