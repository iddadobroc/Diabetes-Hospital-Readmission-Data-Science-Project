```python
import polars as pl
import pyarrow
```

```python
df = pl.read_csv("./../data/diabetic_data.csv")
lazy_df = df.lazy()
```

```python
ctx = pl.SQLContext(diabetic = lazy_df)
```

```python
lazy_result = ctx.execute(
    "SELECT encounter_id, race, age FROM diabetic",
    eager=False,
)

print(lazy_result.explain(optimized=True))  # Mostra o plano, sem executar
result = lazy_result.collect()              # Executa e devolve um DataFrame
```
